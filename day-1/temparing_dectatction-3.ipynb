```python
import hashlib
import time

# Blockchain
blockchain = []
```

```python
# Create hash
def calculate_hash(index, timestamp, data, previous_hash):
    block_string = str(index) + str(timestamp) + str(data) + str(previous_hash)
    return hashlib.sha256(block_string.encode()).hexdigest()
```

```python
# Create a block
def create_block(data, previous_hash):
    index = len(blockchain)
    timestamp = time.time()

    block_hash = calculate_hash(
        index,
        timestamp,
        data,
        previous_hash
    )

    return {
        "index": index,
        "timestamp": timestamp,
        "data": data,
        "previous_hash": previous_hash,
        "hash": block_hash
    }
```

```python
# Add block
def add_block(data):
    if len(blockchain) == 0:
        previous_hash = "0000"
    else:
        previous_hash = blockchain[-1]["hash"]

    block = create_block(data, previous_hash)
    blockchain.append(block)
```

```python
# Display blockchain
def show_blockchain():
    for block in blockchain:
        print("\n--------------------")
        print("Block:", block["index"])
        print("Data:", block["data"])
        print("Previous Hash:", block["previous_hash"])
        print("Hash:", block["hash"])
```

```python

# Verify blockchain
def verify_blockchain():

    for i in range(1, len(blockchain)):

        current_block = blockchain[i]
        previous_block = blockchain[i - 1]

        # Check previous hash
        if current_block["previous_hash"] != previous_block["hash"]:
            return False

        # Recalculate current hash
        new_hash = calculate_hash(
            current_block["index"],
            current_block["timestamp"],
            current_block["data"],
            current_block["previous_hash"]
        )

        if current_block["hash"] != new_hash:
            return False

    return True
```

```python
# Add some blocks
add_block("Alice pays Bob ₹500")
add_block("Bob pays Charlie ₹200")
add_block("Charlie pays David ₹100")

# Display blockchain
show_blockchain()

# Check blockchain
print("\nBlockchain Valid:", verify_blockchain())


# Tampering
print("\n--- Tampering with Block 1 ---")

blockchain[1]["data"] = "Bob pays Charlie ₹5000"

# Check again
print("Blockchain Valid:", verify_blockchain())
```
