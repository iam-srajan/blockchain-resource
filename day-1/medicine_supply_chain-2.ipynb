```
import hashlib
import json
import time
```

```
def calculate_hash(data):
    text = json.dumps(data, sort_keys=True)
    return hashlib.sha256(text.encode("utf-8")).hexdigest()
```

```
def create_block(data, previous_hash):
    block = {
        "data": data,
        "previous_hash": previous_hash,
        "timestamp": int(time.time())
    }

    block["hash"] = calculate_hash(block)
    return block
```

```
def is_valid_chain(blockchain):
    for i in range(1, len(blockchain)):
        current_block = blockchain[i]
        previous_block = blockchain[i - 1]

        block_data = {
            "data": current_block["data"],
            "previous_hash": current_block["previous_hash"],
            "timestamp": current_block["timestamp"]
        }

        if current_block["hash"] != calculate_hash(block_data):
            return False

        if current_block["previous_hash"] != previous_block["hash"]:
            return False

    return True
```

```
genesis_data = {
    "batch": "101",
    "company": "Company A - Manufacturer",
    "event": "Medicine manufactured",
    "time": "9:00 AM"
}

genesis_block = create_block(genesis_data, "0")
blockchain = [genesis_block]
```

```
supply_chain = [
    {
        "batch": "101",
        "company": "Company B - Transporter",
        "event": "Medicine picked up",
        "time": "11:00 AM"
    },
    {
        "batch": "101",
        "company": "Company C - Warehouse",
        "event": "Medicine received",
        "time": "4:00 PM"
    },
    {
        "batch": "101",
        "company": "Company D - Distributor",
        "event": "Medicine distributed",
        "time": "10:00 AM next day"
    },
    {
        "batch": "101",
        "company": "Company E - Pharmacy",
        "event": "Medicine received",
        "time": "5:00 PM next day"
    }
]

for data in supply_chain:
    block = create_block(data, blockchain[-1]["hash"])
    blockchain.append(block)

print("========== BLOCKCHAIN ==========")
print(json.dumps(blockchain, indent=2))

print("\n========== VALIDATION ==========")
print("Blockchain is VALID." if is_valid_chain(blockchain) else "Blockchain is INVALID.")
```

```
print("========== TAMPERING ==========")

blockchain[2]["data"]["event"] = "Medicine received at 2:00 PM"

print("Modified Block 3:")
print(blockchain[2]["data"]["event"])

print("\n========== VALIDATION AFTER TAMPERING ==========")
print("Blockchain is VALID." if is_valid_chain(blockchain) else "Blockchain is INVALID.")
```
