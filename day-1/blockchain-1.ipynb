```python
import hashlib
from time import time
import json
```

```python
def sha256(input):
    return hashlib.sha256(input.encode("utf-8")).hexdigest()
```

```python
hashlib.sha256("hello world".encode()).hexdigest()
```

```python
hashlib.sha256("helloworld".encode()).hexdigest()
```

```python
def createBlock(data, prev_hash):
    block = {
      "data": data,
      "timestamp": int(time()),
      "prev_hash": prev_hash
    }
    block["hash"] = sha256(json.dumps(block, separators=(",", ":")))

    print(json.dumps(block, separators=(",", ":")))
   
    return block
createBlock("hello world", 0)
```

```python
gensis_block = createBlock("gensis block", 0)

blockchain = [gensis_block]
```

```python
block1 = createBlock("block1", blockchain[-1]["hash"])
blockchain.append(block1)

block2 = createBlock("block2", blockchain[-1]["hash"])
blockchain.append(block2)

print(json.dumps(blockchain, indent=2))
```

```python
for i in range(0, 5):
    user_input = input(f"enter data for block {i} ")
    block = createBlock(user_input, blockchain[-1]["hash"])
    blockchain.append(block)


print(json.dumps(blockchain, indent=2))
```
