```python
import hashlib
import time
blockchain = []
```

```python
# Create certificate hash
def calculate_hash(index, timestamp, student, course, certificate_id, previous_hash):

    data = (
        str(index)
        + str(timestamp)
        + student
        + course
        + certificate_id
        + previous_hash
    )

    return hashlib.sha256(data.encode()).hexdigest()

```

```python
# Issue certificate
def issue_certificate(student, course, certificate_id):

    index = len(blockchain)
    timestamp = time.time()

    if len(blockchain) == 0:
        previous_hash = "0000"
    else:
        previous_hash = blockchain[-1]["hash"]

    certificate_hash = calculate_hash(
        index,
        timestamp,
        student,
        course,
        certificate_id,
        previous_hash
    )

    block = {
        "index": index,
        "timestamp": timestamp,
        "student": student,
        "course": course,
        "certificate_id": certificate_id,
        "previous_hash": previous_hash,
        "hash": certificate_hash
    }

    blockchain.append(block)

    print("\nCertificate issued successfully!")
```

```python
# Display certificates
def show_certificates():

    for block in blockchain:

        print("\n-------------------------")
        print("Certificate ID:", block["certificate_id"])
        print("Student:", block["student"])
        print("Course:", block["course"])
        print("Certificate Hash:", block["hash"])
```

```python
# Verify certificate
def verify_certificate(certificate_id):

    for block in blockchain:

        if block["certificate_id"] == certificate_id:

            new_hash = calculate_hash(
                block["index"],
                block["timestamp"],
                block["student"],
                block["course"],
                block["certificate_id"],
                block["previous_hash"]
            )

            if new_hash == block["hash"]:
                print("\nCertificate is VALID")
            else:
                print("\nCertificate has been TAMPERED")

            return

    print("\nCertificate not found")
```

```python
# Issue certificates
issue_certificate(
    "Rahul",
    "Blockchain",
    "CERT101"
)

issue_certificate(
    "Priya",
    "Python",
    "CERT102"
)

# Show certificates
show_certificates()

# Verify certificate
verify_certificate("CERT101")


# Tampering
print("\n--- Someone changes Rahul's certificate ---")

blockchain[0]["student"] = "Amit"

# Verify again
verify_certificate("CERT101")
```
